```python
df=df.drop("customerID",axis=True)
```

```python
df["TotalCharges"] = pd.to_numeric(
    df["TotalCharges"],
    errors="coerce"
)
```

```python
df = df.drop(columns=["tenure_group"], errors="ignore")
```

```python
df["Churn"] = df["Churn"].map({
    "No": 0,
    "Yes": 1
})
```

```python
X = df.drop(columns=["Churn"])
y = df["Churn"]
```

```python
numerical_features = X.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

categorical_features = X.select_dtypes(
    include=["object", "str"]
).columns.tolist()

print("Numerical features:")
print(numerical_features)

print("\nCategorical features:")
print(categorical_features)
```

```python

from sklearn.model_selection import train_test_split

# Keep the test set untouched until final evaluation
X_trainval, X_test, y_trainval, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# Split the remaining data into training and validation sets
X_train, X_val, y_train, y_val = train_test_split(
    X_trainval,
    y_trainval,
    test_size=0.20,
    random_state=42,
    stratify=y_trainval
)

print("Training set:", X_train.shape)
print("Validation set:", X_val.shape)
print("Test set:", X_test.shape)

print("\nChurn rate:")
print("Train:", round(y_train.mean(), 3))
print("Validation:", round(y_val.mean(), 3))
print("Test:", round(y_test.mean(), 3))
```

```python
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
numeric_pipeline=Pipeline([
    ("imputer",SimpleImputer(strategy="median")),
    ("scaler",StandardScaler())
])
from sklearn.preprocessing import OneHotEncoder

categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(
        handle_unknown="ignore",
        sparse_output=False
    ))
])
from sklearn.compose import ColumnTransformer

preprocessor = ColumnTransformer([
    ("num", numeric_pipeline, numerical_features),
    ("cat", categorical_pipeline, categorical_features)
])
```

```python

# Fit preprocessing ONLY on training data
X_train_processed = preprocessor.fit_transform(X_train)

# Transform validation and test data without fitting again
X_val_processed = preprocessor.transform(X_val)
X_test_processed = preprocessor.transform(X_test)

print("Train:", X_train_processed.shape)
print("Validation:", X_val_processed.shape)
print("Test:", X_test_processed.shape)

print("Train NaN:", np.isnan(X_train_processed).sum())
print("Validation NaN:", np.isnan(X_val_processed).sum())
print("Test NaN:", np.isnan(X_test_processed).sum())
```

```python

from tensorflow.keras import Sequential, Input
from tensorflow.keras.layers import Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping

model = Sequential([
    Input(shape=(X_train_processed.shape[1],)),

    Dense(64, activation="relu"),
    Dropout(0.2),

    Dense(32, activation="relu"),
    Dropout(0.2),

    Dense(1, activation="sigmoid")
])

model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy", "Precision", "Recall", "AUC"]
)

early_stopping = EarlyStopping(
    monitor="val_loss",
    patience=5,
    restore_best_weights=True
)

history = model.fit(
    X_train_processed,
    y_train,
    validation_data=(X_val_processed, y_val),
    epochs=50,
    batch_size=32,
    callbacks=[early_stopping],
    verbose=1
)
```

```python

from sklearn.utils.class_weight import compute_class_weight
import numpy as np

classes = np.unique(y_train)

weights = compute_class_weight(
    class_weight="balanced",
    classes=classes,
    y=y_train
)

class_weights = dict(zip(classes, weights))
print("Class weights:", class_weights)

history_weighted = model.fit(
    X_train_processed,
    y_train,
    validation_data=(X_val_processed, y_val),
    epochs=50,
    batch_size=32,
    callbacks=[early_stopping],
    class_weight=class_weights,
    verbose=1
)
```

```python

import numpy as np
import pandas as pd

from sklearn.metrics import (
    precision_score,
    recall_score,
    f1_score
)

# Step 1: Validation data par churn probabilities predict karein
# Model har customer ke liye 0 aur 1 ke darmiyan probability dega
val_prob_weighted = model.predict(
    X_val_processed,
    verbose=0
).ravel()

# Step 2: Different thresholds test karein
# Lower threshold = zyada customers churn predict ho sakte hain
thresholds = np.arange(0.15, 0.56, 0.025)

results = []

for threshold in thresholds:

    # Probability threshold se zyada ya equal ho to Churn = 1
    # Warna No Churn = 0
    val_pred = (val_prob_weighted >= threshold).astype(int)

    # Har threshold par performance calculate karein
    results.append({
        "Threshold": round(float(threshold), 3),

        # Predicted churn customers mein se kitne actually churn hue?
        "Precision": precision_score(
            y_val, val_pred, zero_division=0
        ),

        # Actual churn customers mein se kitne identify hue?
        "Recall": recall_score(
            y_val, val_pred, zero_division=0
        ),

        # Precision aur recall ka combined score
        "F1": f1_score(
            y_val, val_pred, zero_division=0
        )
    })

# Step 3: Results ko table mein display karein
threshold_df = pd.DataFrame(results)

print(threshold_df.round(3).to_string(index=False))
```

```python

# Business goal: validation recall at least 80% ho
target_recall = 0.80

# Sirf woh thresholds select karein jo 80% recall achieve karte hain
eligible_thresholds = threshold_df[
    threshold_df["Recall"] >= target_recall
]

if not eligible_thresholds.empty:

    # Eligible thresholds mein se highest precision wala choose karein
    # Agar precision same ho, to higher F1 ko preference milegi
    best_row = eligible_thresholds.sort_values(
        ["Precision", "F1"],
        ascending=False
    ).iloc[0]

    best_threshold_weighted = float(best_row["Threshold"])

    print("Selected threshold:", best_threshold_weighted)
    print("\nSelected validation results:")
    print(best_row)

else:
    print(
        "80% recall achieve nahi hua. "
        "Threshold table check karein."
    )
```

```python

# Step 1: Evaluation ke liye required metrics import karein
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    average_precision_score
)
import matplotlib.pyplot as plt

# Step 2: Validation data se select kiya hua threshold use karein
# Test results dekh kar threshold change nahi karna
best_threshold_weighted = 0.45

# Step 3: Test customers ki churn probabilities predict karein
# Model har customer ke churn hone ki probability deta hai
test_prob_weighted = model.predict(
    X_test_processed,
    verbose=0
).ravel()

# Step 4: Probabilities ko 0 (No Churn) aur 1 (Churn) mein convert karein
# Probability >= 0.45 ho to customer ko Churn predict karein
test_pred_weighted = (
    test_prob_weighted >= best_threshold_weighted
).astype(int)

# Step 5: Test set par performance metrics calculate karein
print("Threshold:", best_threshold_weighted)
print("Accuracy :", accuracy_score(y_test, test_pred_weighted))
print("Precision:", precision_score(y_test, test_pred_weighted))
print("Recall   :", recall_score(y_test, test_pred_weighted))
print("F1-score :", f1_score(y_test, test_pred_weighted))
print("ROC-AUC  :", roc_auc_score(y_test, test_prob_weighted))
print("PR-AUC   :", average_precision_score(y_test, test_prob_weighted))

# Step 6: Har class ki detailed performance dekhein
print("\nClassification Report:")
print(classification_report(
    y_test,
    test_pred_weighted,
    target_names=["No Churn", "Churn"],
    zero_division=0
))

# Step 7: Confusion matrix se correct aur incorrect predictions dekhein
cm_weighted = confusion_matrix(y_test, test_pred_weighted)

display = ConfusionMatrixDisplay(
    confusion_matrix=cm_weighted,
    display_labels=["No Churn", "Churn"]
)

display.plot()
plt.title("Weighted ANN - Final Test Confusion Matrix")
plt.show()
```

```python

import matplotlib.pyplot as plt

# Training aur validation loss ko history se retrieve karein
train_loss = history_weighted.history["loss"]
val_loss = history_weighted.history["val_loss"]

# Har epoch ki loss values plot karein
plt.figure(figsize=(9, 5))

plt.plot(train_loss, label="Training Loss")
plt.plot(val_loss, label="Validation Loss")

plt.xlabel("Epoch")
plt.ylabel("Binary Cross-Entropy Loss")
plt.title("Weighted ANN: Training vs Validation Loss")

plt.legend()
plt.grid(True)
plt.show()
```
